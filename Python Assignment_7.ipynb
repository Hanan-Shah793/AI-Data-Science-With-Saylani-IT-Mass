# Student Score Validation and Ranking System :

```python
lines = [
"Ayesha, 78, 91, 65",
"  bilal ,100,80",
"Chen, 88, 90",
"Dana",
"Omar, 85, abc",
"",
]


new_list = []

for line in lines:
    if line.strip() == "":
        continue
    pieces = line.split(",")
    if len(pieces) < 2:
        continue
    name_clean = pieces[0].strip().title()
    
    score =[]
    valid = True
    
    for piece in pieces[1:]:
        piece = piece.strip()
        
        if not piece.isdigit():
            valid = False
            break
        
        score.append(int(piece))
        
    if not valid:
        continue
        
    avg = sum(score) / len(score)
    avg = round(avg , 1)
        
    new_list.append([avg , name_clean])
        
new_list.sort()
new_list.reverse()
        
        
rank = 1

for list in new_list:
    print(f"{rank}. {list[1]} - {list[0]:.1f}")
    rank += 1
    
        

    
```
