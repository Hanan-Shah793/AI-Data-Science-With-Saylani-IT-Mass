#caesar_cipher.py

```python
messages = ["Hello, World!", "Khoor, Zruog!", "abc"]
shifts = [3, -3, 29]
alphabet = "abcdefghijklmnopqrstuvwxyz"


for messages , shifts in zip(messages , shifts):
    answer = ""
    
    shifts = shifts % 26
    
    for letter in messages:
        if letter.isalpha():
            position = alphabet.index(letter.lower())
            new_position = (position + shifts) % 26
            new_letter = alphabet[new_position]
            
            if letter.isupper():
                new_letter = new_letter.upper()
                
            answer = answer + new_letter
            
        else:
            answer = answer + letter
    print(answer)     
```
