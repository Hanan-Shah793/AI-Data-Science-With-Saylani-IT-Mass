# Customer Feedback Text Analyzer :

```python
user = input("enter your feedback:")

user_strip = user.strip()
for_lower_case = user.lower()
for_white_spaces = user.split()

Character_Count = len(user)
word_count = user.split()
vowels = 0
for char in user:
    if char in "aeiouAEIOU":
        vowels += 1
spaces = for_lower_case.count(" ")
good_count = for_lower_case.count("good")
bad_count = for_lower_case.count("bad")

if good_count > 0 and good_count == 0:
    category = "Positive"

elif bad_count > 0 and bad_count == 0:
    category = "Negative"
    
elif good_count > 0 and bad_count > 0:
    category = "Not a good \n Not a bad"
    
    
else:
    category = "Netural"
    
longest_word = ""
for word in word_count:
    if len(word) > len(longest_word):
        longest_word = word
        
print("the feedback we obtain form user is: " , user)
print("feedback strip:" , user_strip)
print("feedback in lower case:" , for_lower_case)
print("for white spaces in feedback: " , for_white_spaces)
print("vowels in feedback is:" , vowels)
print("feedback category is:" , category)
print("longest word in feedback is:" , longest_word)
```
