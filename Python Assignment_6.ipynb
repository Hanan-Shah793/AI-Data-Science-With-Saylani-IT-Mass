# Merge Sorted Lists Without Sorting :

```python
left = [1, 4, 7, 7]
right = [2, 4, 9]

result = []

i = 0 
j = 0

while i < len(left) and j < len(right):
    if left[i] < right[j]:
        result.append(left[i])
        i += 1
    
    else:
        result.append(right[j])
        j += 1
        
    
while i < len(left):
    result.append(left[i])
    i += 1
    
    
while j < len(right):
    result.append(right[j])
    j += 1
        
print(result)
```
