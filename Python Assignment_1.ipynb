#String Cleaning and Case Conversion

```python
raw_names = ["  aYESHA   khan ", "JOHN   smith", "mary  jane  WATSON"]
for name in raw_names:
    extra_spaces = name.split()
    upper_case = name.upper()
    lower_case = name.lower()
    first_capital = name.title()
    print("removing extra spaces from start to end:" , extra_spaces)
    print("text in upper case:", upper_case)
    print("text in lower case:" , lower_case)
    print("for first letter capital:" , first_capital) 
```
