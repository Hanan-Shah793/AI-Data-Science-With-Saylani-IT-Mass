#Longest Consecutive Repeating Characters:


```python
samples = ["aabbbcc", "xxyy", "mississippi", "abc"]

for sample in samples:
    current_count = 1
    longest_count = 1
    
    current_repeat = sample[0]
    longest_repeat = sample[0]


    for i in range(1 , len(sample)):
        if sample[i] == sample[i -1]:
            current_count += 1
        else:
            current_count = 1
        
        if current_count > longest_count:
            longest_count = current_count
            longest_repeat = sample[i] * current_count
    print(longest_repeat , "(", longest_count , ")" , sep="")
    
```
