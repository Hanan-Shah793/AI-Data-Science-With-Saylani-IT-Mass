# Weekly Internet Data Usage Analyzer :

```python
days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]

usage = []

for i in range(7):
    day = days[i]
    GBs = int(input("enter the GBS:"+ day + ": "))
    usage.append(GBs)
    
print("\nDATA USAGE REPORT")  
total = 0
   
for i in range(7):
    day = days[i]
    print(day, ":" , usage[i] , "GB")
    total += usage[i]
    
    
avg_daily_usage = total / len(usage)
print("total usage is:" , total , "GB")
print("avg daily usage is:" , avg_daily_usage , "GB")
    
highest = usage[0]
for i  in range(1 , len(usage)):
    if usage[i] > highest:
        highest = usage[i]
print("Highest usage is:" , highest , "GB")
        
heavy_days_usage = 0
for i in range(0 , len(usage)):
    if usage[i] > 4:
        heavy_days_usage += 1
print("heavy usage of day count is:" , heavy_days_usage , "days")

if total >= 0 and total <= 15 :
    print("category = light")
elif total >= 15 and total <= 30 :
    print("category = Moderate")
elif total >= 30 :
    print("category = heavy")
```
