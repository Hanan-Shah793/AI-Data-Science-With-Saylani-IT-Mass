# Supermarket Billing and Discount System :

```python
n_Products = int(input("How many products you want to enter:"))
sub_total = 0
print("-----SUPER_MART------")

for i in range(n_Products):
    product_name = input("enter the name:")
    product_price = float(input("enter the price:"))
    product_quantity = int(input("enter the quantity:"))
    total = product_price * product_quantity
    sub_total += total
    
    

    print("product name is:" , product_name)
    print("product price is:" , product_price)
    print("quantity is:" , product_quantity)
    print("product total is:" , total)
  
    
if sub_total >=10000:
    discount = sub_total * 10 /100
        
elif sub_total > 5000 and sub_total <= 9999:
    discount = sub_total * 5 /100
        
else:
    discount = 0
    print("NO Discount")
    

final_amount = sub_total - discount
    
    

print("subtotal of all products is:" , sub_total)
print("discount we give you:" , discount)
print("final amount is:" , final_amount) 
```
